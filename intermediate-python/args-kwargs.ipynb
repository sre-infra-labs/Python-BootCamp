# Force keyword arguments

```python
def foo(*, a, b, c):
    print(a, b, c)

# correct
foo(a=1, b=2, c=3)

print('--'*20)

# error
foo(1, 2, 3)
```

# Argument Unpacking

```python
def foo(a, b, c):
    print(a, b, c)

nums = [1, 2, 3]

foo(*nums)
```

```python
def foo(a, b, c):
    print(a, b, c)

nums = [1, 2, 3, 4]

foo(*nums[:3:])
```

```python
def foo(a, b, c):
    print(a, b, c)

my_dict = {'a':1, 'b':2, 'c':3}

foo(**my_dict)
```

```python
numbers = [1, 2, 3, 4, 5, 6]

first, *middle, last = numbers

print(first)
print(middle)
print(*middle)
print(last)
```

# Global Variable in Local scope

```python
def foo():
    global number
    number = 2
    print(f"number inside function: {number}")

number = 0
foo()

print(number)
```

# Call by Object Reference

2 Rules -

- Parameters passed in are actually reference to an object, but the reference is passed by value
- There is difference b/w mutuable (list, dict) & immutable (int, str, set) data types

### Immutable Object argument (int)

```python
def foo(x):
    x = 5

var = 10
foo(var)

print(var)

# Immutable objects cannot be changed. So inside foo() a local variable is created instead of referencing the original variable.
```

### Mutable object argument (list)

```python
def foo(a_list):
    a_list.append(4)
    a_list[0] = -100

my_list = [1, 2, 3]
foo(my_list)

print(my_list)

# Mutable objects allow change. So inside foo() the same passed variable is used.
```

### Assignment operator with Mutable object argument

```python
def foo_a(list_a):
    list_a = list_a + [200, 300, 400]

list_a = [1, 2, 3]
foo_a(list_a)
print(list_a)

print()
print ('--'*20)
print()

def foo_b(list_b):
    list_b += [200, 300, 400]

list_b = [1, 2, 3]
foo_b(list_b)
print(list_b)
```

# Flexible Functions: `*args` and `**kwargs`

### Example 01: Using `*args` and `**kwargs` with decorator

In below example, we are using debug `decorator` to debug function call.

Note, even dynamic arguments are replace with static values.

```python
import functools

def debug(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        args_repr = [repr(a) for a in args]
        kwargs_repr = [f"{k}={v!r}" for k,v in kwargs.items()]
        signature = ", ".join(args_repr + kwargs_repr)
        print(f"DEBUG: Calling {func.__name__}({signature})")
        result = func(*args, **kwargs)
        # print(f"{func.__name__!r} returned {result!r}")
        return result
    return wrapper

@debug
def apply_operator(operator, *operands):
    """Applies operator to a variable number of operands. Supports 'add' and 'mul'.

    Args:
        operator (str): The operator to apply. Must be either 'add' or 'mul'.
        *operands (int or float): Zero or more numbers to be combined.

    Returns:
        int or float: The result of applying the operator on the operands.

    Raises:
        ValueError: Raised when operator is not 'add' nor 'mul'.
    """

    if operator == 'add':
        result = sum(operands)
    elif operator == 'mul':
        result = 1
        for n in operands:
            result *= n
    else:
        raise ValueError(f"Unknown operator '{operator}'. Supported values are 'add' & 'mul'")

    return result

my_nums = [1,2,3]

# print(apply_operator('add', 1,2,3,4))
print(apply_operator('add', *my_nums,4))

# print(apply_operator('add', 1,2,3,4,5,6))
print(apply_operator('add', *(my_nums+[4,5,6])))

print(apply_operator('mul', *my_nums))

print(apply_operator('mul', 1,2,3,10))

print(apply_operator('div', 1,2,3))
        
```

### Example 02: Using `*args` and `**kwargs` with positional arguments



```python
def foo(a, b, *args, **kwargs):
    print(a, b)
    for arg in args:
        print(arg)
    for key in kwargs:
        print(key, kwargs[key])

foo(1, 2, 3, 4, 5, six=6, seven=7)
```
